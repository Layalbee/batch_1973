**Cleaning data**

```python
df_tournament_filter['nb_teams'] = (
    df_tournament_filter['nb_teams']
    .astype(str)  # Ensure all values are strings
    .str.replace(r' Team[s]?$', '', regex=True)  # Handle both "Team" and "Teams"
    .replace('nan', None)
    .pipe(pd.to_numeric, errors='coerce')  # Convert to numbers
    .fillna(0)
    .astype(int)
)
```

```python
df_tournament_filter['cash_prize'] = (
    df_tournament_filter['cash_prize']
    .astype(str)                              # Ensure all values are strings
    .str.replace('[\$,]', '', regex=True)     # Remove $ and commas
    .replace('nan', None)                     # Optional: Handle "nan" strings
    .pipe(pd.to_numeric, errors='coerce')     # Convert to float
    .fillna(0)                                 # Handle NaNs if needed
    .astype(int)
)
```

```python
df_tournament_filter['nb_players'] = (
    df_tournament_filter['nb_players']
    .astype(str)
    .str.replace(r' Player[s]?$', '', regex=True)  # Removes " Player" or " Players"
    .replace('nan', None)
    .pipe(pd.to_numeric, errors='coerce')
    .fillna(0)
    .astype(int)
)
```
